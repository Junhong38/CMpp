```python
def knn(x, k):
    inner = -2*torch.matmul(x.transpose(2, 1), x)
    xx = torch.sum(x**2, dim=1, keepdim=True)
    pairwise_distance = -xx - inner - xx.transpose(2, 1)
 
    idx = pairwise_distance.topk(k=k, dim=-1)[1]   # (batch_size, num_points, k)
    return idx
```

```python
def compute_directional_vectors_pytorch(x, k=512):
    x = x.transpose(1, 2)  # (1, 3, num_points)

    # KNN 인덱스 계산
    idx = knn(x, k)  # (1, num_points, k)
    
    # 이웃 포인트 좌표 추출
    batch_size, _, num_points = x.size()
    idx_base = torch.arange(0, batch_size, device=x.device).view(-1, 1, 1) * num_points
    idx = idx + idx_base
    idx = idx.view(-1)
    
    x = x.transpose(1, 2).contiguous()  # (1, num_points, 3)
    feature = x.view(batch_size * num_points, -1)[idx, :]
    feature = feature.view(batch_size, num_points, k, -1)  # (1, num_points, k, 3)
    
    # 중심 포인트와의 차이 벡터 계산
    x = x.view(batch_size, num_points, 1, -1)
    vectors = x - feature  # (1, num_points, k, 3)
    
    # 벡터 합산
    summed_vectors = torch.sum(vectors, dim=2)  # (1, num_points, 3)
    
    # 벡터 정규화
    norms = torch.norm(summed_vectors, dim=2, keepdim=True)
    normalized_vectors = summed_vectors / (norms + 1e-8)  # (1, num_points, 3)
    
    return normalized_vectors  # (num_points, 3)
```

```python
def compute_directional_vectors(src_pcd_raw, k=512):
    tree = KDTree(src_pcd_raw)
    distances, indices = tree.query(src_pcd_raw, k=k+1)
    neighbor_indices = indices[:, 1:]
    
    # 이웃 벡터 계산
    vectors = src_pcd_raw[:, np.newaxis, :] - src_pcd_raw[neighbor_indices]
    summed_vectors = np.sum(vectors, axis=1)
    
    # 벡터 정규화
    norms = np.linalg.norm(summed_vectors, axis=1, keepdims=True)
    normalized_vectors = summed_vectors / norms
    
    return normalized_vectors
```

```python
def find_k_nearest_neighbors(point_cloud, idx, k):
    tree = KDTree(point_cloud)
    distances, indices = tree.query(point_cloud[idx], k=k+1)  # k+1은 자기 자신을 제외하기 위해
    return indices[1:]  # 첫 번째는 자기 자신이므로 제외
```

```python
vectors_per_point = []

for point_idx in range(len(src_pcd_raw)):
    knn_indices = find_k_nearest_neighbors(src_pcd_raw, point_idx, 512)
    vectors = []
    for i in knn_indices:
        vectors.append(src_pcd_raw[point_idx] - src_pcd_raw[i])
    # Sum the vectors for this point
    summed_vector = np.sum(vectors, axis=0)
    vectors_per_point.append(summed_vector)
vectors_per_point = np.array(vectors_per_point)
vectors_per_point = vectors_per_point / np.linalg.norm(vectors_per_point, axis=1)[:, np.newaxis]
```

```python
def _is_trg_larger(src_pcd, trg_pcd):
    src_volume = np.prod(src_pcd.max(axis=0) - src_pcd.min(axis=0))
    trg_volume = np.prod(trg_pcd.max(axis=0) - trg_pcd.min(axis=0))

    return src_volume < trg_volume

def _pairwise_mating(src_pcd, trg_pcd, rotat, trans, is_trg_larger):
    pcd_t = []
    if is_trg_larger:
        src_pcd_t = _transform(src_pcd, rotat, -trans, True)
        pcd_t = [src_pcd_t, trg_pcd]
    else:
        trg_pcd_t = _transform(trg_pcd, np.linalg.inv(rotat), trans, False)
        pcd_t = [src_pcd, trg_pcd_t]
    return np.concatenate(pcd_t, axis=0), pcd_t

def _transform(pcd, rotat=None, trans=None, rotate_first=True):
    if rotat is None:
        rotat = np.eye(3)
    if trans is None:
        trans = np.zeros(3)

    if rotate_first:
        return np.dot(pcd, rotat.T) + trans
    else:
        return np.dot(pcd + trans, rotat.T)
```

```python
is_trg_larger = _is_trg_larger(src_pcd, trg_pcd)

# Assemble using prediction, pseudo-gt, and ground-truth
assm_pred, pcds_pred = _pairwise_mating(src_pcd, trg_pcd, estimated_rotat, estimated_trans, is_trg_larger)
```
