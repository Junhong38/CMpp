```python
import pandas as pd
```

```python
column_headers = ['idx', 'class', 'src_pcd', 'trg_pcd', 'rrmse', 'trmse', 'crd', 'cd', 'c_loss', 'occ_loss', 'p_loss', 'o_loss']

v7 = pd.read_csv('v7_small.txt', sep = "|",  header=None, names=column_headers)
unet_v1 = pd.read_csv('unet_v1_small.txt', sep = "|",  header=None, names=column_headers)
```

```python
unet_v1
```

```python
df_diff = v7.copy()
for col in v7.columns[4:]:  # Skip 'idx' column
    df_diff[col] = v7[col] - unet_v1[col]
```

```python
class_metrics = df_diff.groupby('class').mean(numeric_only=True)
```

```python
class_metrics
```
